```python

## Download this file: https://github.com/cwrussell/bioscreen/blob/master/bioscreen.py

bioscreen_py_directory = "C:/Users/c2006576/Downloads"

bioscreen_data_csv = "C:/Users/c2006576/Downloads/E.coli CM growth.csv"
bioscreen_config = "C:/Users/c2006576/Downloads/data.config.txt"

output_directory = "C:/Users/c2006576/Downloads/"

remove_labels = True
colour_palette = "jet"
```

```python
## Install Bioscreen.py from Github: https://github.com/cwrussell/bioscreen/blob/master/bioscreen.py
import sys
import pandas as pd 
sys.path.insert(0, bioscreen_py_directory)

import bioscreen

## Configure config file (if saved as txt from excel)
with open(bioscreen_config, "r", encoding="cp1252") as f:
    content = f.read()

content = content.replace("–", "-")

with open(bioscreen_config, "w", encoding="cp1252") as f:
    f.write(content)

## Set up Bioscreen Environment.
expt = bioscreen.Experiment()
expt.set_config_from_file(bioscreen_config)

## Create Summaries and write file.
expt.summarize(bioscreen_data_csv, input_encoding="cp1252")
expt.write_summary(output_directory + "Bioscreen_summary.csv")

if remove_labels == True:
    ## Rename plotting labels to remove group ID (Optional)

    expt.summary_data.columns = [
        col.replace('CaTEST_', '') if isinstance(col, str) else col
        for col in expt.summary_data.columns
    ]

    expt.summary_data.columns = [
        col.lstrip('_') if isinstance(col, str) and col != 'Time' else col
        for col in expt.summary_data.columns
    ]

else:
    pass

expt.graph(output_directory + "Bioscreen_plot.png", line_colors=colour_palette)
```
